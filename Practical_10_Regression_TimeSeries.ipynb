# Practical 10: Regression Analysis and Time Series Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# 1. Load dataset
df = pd.read_csv("sales_data.csv")

# 2. Display first five records
print("First Five Records:")
print(df.head())

# 3 & 4. Identify variables
X = df[["Advertising"]]
y = df["Sales"]
print("\n3. Independent Variable: Advertising")
print("Dependent Variable: Sales")

# 5. Check missing values
print("\n5. Missing Values:")
print(df.isnull().sum())

# 6. Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["Advertising"], df["Sales"])
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Advertising vs Sales")
plt.show()

# 7. Develop regression model
model = LinearRegression()
model.fit(X, y)

# 8. Regression equation
b0 = model.intercept_
b1 = model.coef_[0]
print("\n8. Regression Equation:")
print(f"Sales = {b0:.2f} + {b1:.2f}(Advertising)")

# 9. Predict sales
advertising_value = [[75]]
predicted_sales = model.predict(advertising_value)
print("\n9. Predicted Sales for Advertising = 75:")
print(predicted_sales[0])

# 10. R² score
predictions = model.predict(X)
r2 = r2_score(y, predictions)
print("\n10. R² Score:", r2)

# 11. MSE
mse = mean_squared_error(y, predictions)
print("\n11. Mean Squared Error:", mse)

# 12. Actual vs predicted
comparison = pd.DataFrame({
    "Actual Sales": y,
    "Predicted Sales": predictions
})
print("\n12. Actual vs Predicted Sales:")
print(comparison)

# 13. Regression line
plt.figure(figsize=(8, 5))
plt.scatter(df["Advertising"], df["Sales"])
plt.plot(df["Advertising"], predictions)
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Regression Line")
plt.show()

# 14. Interpretation
print("\n14. Interpretation:")
print("The regression model shows the relationship between advertising expenditure and sales.")

# 15. Conclusion
print("\n15. Conclusion:")
print("Sales generally increase as advertising expenditure increases.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error

# 16. Load monthly sales data
df = pd.read_csv("monthly_sales.csv")
print("Monthly Sales Data:")
print(df)

# 17. Convert date column
df["Date"] = pd.date_range(start="2025-01-01", periods=len(df), freq="MS")
df.set_index("Date", inplace=True)

# 18. Plot monthly sales
plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Sales"], marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid()
plt.show()

# 19. Observe trend
print("\n19. Trend:")
print("Sales generally show an increasing trend over the months.")

# 20. Exponential Smoothing
model_es = ExponentialSmoothing(df["Sales"], trend="add").fit()

# 21. Forecast next 3 months
forecast_es = model_es.forecast(3)
print("\n21. Exponential Smoothing Forecast:")
print(forecast_es)

# 22. Plot actual and forecast
plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Sales"], label="Actual")
plt.plot(forecast_es.index, forecast_es, label="Forecast")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Exponential Smoothing Forecast")
plt.legend()
plt.show()

# 23. Forecasting error
fitted_es = model_es.fittedvalues
mse_es = mean_squared_error(df["Sales"], fitted_es)
print("\n23. Exponential Smoothing MSE:", mse_es)

# 24. ARIMA
model_arima = ARIMA(df["Sales"], order=(1, 1, 1)).fit()

# 25. ARIMA forecast
forecast_arima = model_arima.forecast(3)
print("\n25. ARIMA Forecast:")
print(forecast_arima)

# 26. Compare forecasts
forecast_comparison = pd.DataFrame({
    "Exponential Smoothing": forecast_es.values,
    "ARIMA": forecast_arima.values
})
print("\n26. Forecast Comparison:")
print(forecast_comparison)

# 27. Compare errors
fitted_arima = model_arima.fittedvalues
mse_arima = mean_squared_error(df["Sales"], fitted_arima)
print("\n27. Forecasting Errors:")
print("Exponential Smoothing MSE:", mse_es)
print("ARIMA MSE:", mse_arima)

# 28. Lower error
better_model = "Exponential Smoothing" if mse_es < mse_arima else "ARIMA"
print("\n28. Model with Lower Forecasting Error:", better_model)

# 29. Conclusion
print("\n29. Conclusion:")
print(f"The {better_model} model gives the lower forecasting error for the given dataset.")
```
