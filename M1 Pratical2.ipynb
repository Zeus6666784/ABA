```python
import pandas as pd
```

```python
df = pd.read_csv("walmart-sales-dataset-of-45stores.csv")
```

```python
print(df.head())
```

```python
sales = df["Weekly_Sales"]
```

```python
print("Mean :", sales.mean())
```

```python
print("Median :", sales.median())
```

```python
print("Mode :")
print(sales.mode())
```

```python
print("Minimum :", sales.min())
```

```python
print("Maximum :", sales.max())
```

```python
print("Range :", sales.max() - sales.min())
```

```python
print("Variance :", sales.var())
```

```python
print("Standard Deviation :", sales.std())
```

```python
print("Count :", sales.count())
```

```python
print("Sum :", sales.sum())
```

```python
print("\nDescriptive Statistics")
print(sales.describe())
```

# Practical 2B

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
plt.hist(sales, bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=sales, color="orange")
plt.title("Box Plot of Sales")
plt.xlabel("Sales")
plt.show()
```

```python
print("Store-wise Transaction Count")
print(df["Store"].value_counts())
```

```python
store_sales = df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
print(store_sales)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
store_sales.plot(kind="bar", color="steelblue")
plt.title("Total Weekly Sales by Store")
plt.xlabel("Store")
plt.ylabel("Total Weekly Sales")
plt.tight_layout()
plt.show()
```
