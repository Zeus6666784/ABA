# Netflix Dataset Analysis
Beginner-level solutions

Dataset file: `NetFlix - NetFlix.csv`

## Q1

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix - NetFlix.csv")

genres = df["listed_in"].dropna().str.split(", ").explode()

genre_count = genres.value_counts()

print("Top 10 Genres")
print(genre_count.head(10))

print("\nBottom 10 Genres")
print(genre_count.tail(10))

print("\nMost Popular Genre:", genre_count.idxmax())
print("Least Popular Genre:", genre_count.idxmin())

genre_count.head(10).plot(kind="bar")
plt.title("Top 10 Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

## Q2

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix - NetFlix.csv")

movies = df[df["type"]=="Movie"]
tv = df[df["type"]=="TV Show"]

old_movie = movies.loc[movies["release_year"].idxmin()]
new_movie = movies.loc[movies["release_year"].idxmax()]
old_tv = tv.loc[tv["release_year"].idxmin()]
new_tv = tv.loc[tv["release_year"].idxmax()]

print("Oldest Movie")
print(old_movie)

print("\nNewest Movie")
print(new_movie)

print("\nOldest TV Show")
print(old_tv)

print("\nNewest TV Show")
print(new_tv)
```

## Q3

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix - NetFlix.csv")

countries = df["country"].dropna().str.split(", ").explode()
country_count = countries.value_counts()

print("Country with Highest Netflix Content:")
print(country_count.idxmax())

print("\nTop 10 Countries")
print(country_count.head(10))

print("\nBottom 10 Countries")
print(country_count.tail(10))

country_count.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Countries")
plt.xlabel("Content Count")
plt.show()
```

## Q4

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix - NetFlix.csv")

directors = df["director"].dropna().str.split(", ").explode()
director_count = directors.value_counts()

movie_director = movies["director"].dropna().str.split(", ").explode().value_counts()
tv_director = tv["director"].dropna().str.split(", ").explode().value_counts()

print("Director with Highest Movies:")
print(movie_director.idxmax())

print("\nDirector with Highest TV Shows:")
print(tv_director.idxmax())

print("\nTop 10 Directors")
print(director_count.head(10))

director_count.head(10).plot(kind="bar")
plt.title("Top 10 Directors")
plt.xticks(rotation=45)
plt.show()
```

## Q5

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix - NetFlix.csv")

before_2020 = len(df[df["release_year"] < 2020])
during_2020 = len(df[df["release_year"] == 2020])
after_2020 = len(df[df["release_year"] > 2020])

labels = ["Before 2020","During 2020","After 2020"]
values = [before_2020,during_2020,after_2020]

print("Before 2020:", before_2020)
print("During 2020:", during_2020)
print("After 2020:", after_2020)

plt.bar(labels, values)
plt.title("Netflix Releases")
plt.ylabel("Number of Titles")
plt.show()
```
