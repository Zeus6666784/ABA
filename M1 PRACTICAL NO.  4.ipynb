```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Simulate 1000 dice rolls
rolls = np.random.randint(1, 7, 1000)

# Display first 20 outcomes
print("First 20 Dice Rolls:")
print(rolls[:20])
```

### 2. Calculate the experimental probability of getting each number from 1 to 6.

```python
frequency = pd.Series(rolls).value_counts().sort_index()
experimental_probability = frequency / 1000

print("Experimental Probability:")
print(experimental_probability)
```

### 3. Calculate the theoretical probability of getting each number from 1 to 6.

```python
theoretical_probability = 1 / 6
print("Theoretical Probability of each number:", theoretical_probability)
```

### 4. Find the probability of getting 1, 2, 3, 4, 5, 6.

```python
print("Probability of getting 1:", np.mean(rolls == 1))
print("Probability of getting 2:", np.mean(rolls == 2))
print("Probability of getting 3:", np.mean(rolls == 3))
print("Probability of getting 4:", np.mean(rolls == 4))
print("Probability of getting 5:", np.mean(rolls == 5))
print("Probability of getting 6:", np.mean(rolls == 6))
```

### 5. Find the probability of getting an even number.

```python
even_probability = np.mean(rolls % 2 == 0)
print("Probability of getting an even number:", even_probability)
```

### 6. Find the probability of getting an odd number.

```python
odd_probability = np.mean(rolls % 2 != 0)
print("Probability of getting an odd number:", odd_probability)
```

### 7. Find the probability of getting a number greater than 4.

```python
probability_greater_4 = np.mean(rolls > 4)
print("Probability of getting a number greater than 4:", probability_greater_4)
```

### 8. Find the probability of getting a number less than 4.

```python
probability_less_4 = np.mean(rolls < 4)
print("Probability of getting a number less than 4:", probability_less_4)
```

### 9. Find the probability of getting a number greater than or equal to 4.

```python
probability_greater_equal_4 = np.mean(rolls >= 4)
print("Probability of getting a number greater than or equal to 4:",
      probability_greater_equal_4)
```

### 10. Create a bar chart showing the frequency of outcomes 1 to 6.

```python
plt.pie(frequency.values, labels=frequency.index, autopct="%1.1f%%")
plt.title("Percentage Distribution of Dice Outcomes")
plt.show()
```

### 13. Create a line chart comparing experimental probability and theoretical probability.

```python
theoretical = [1/6] * 6

plt.plot(range(1, 7), experimental_probability, marker="o",
         label="Experimental")
plt.plot(range(1, 7), theoretical, marker="o",
         label="Theoretical")

plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.title("Experimental vs Theoretical Probability")
plt.legend()
plt.show()
```

### 14. Create a chart showing the frequency of Even vs Odd outcomes.

```python
even_frequency = np.sum(rolls % 2 == 0)
odd_frequency = np.sum(rolls % 2 != 0)

plt.bar(["Even", "Odd"], [even_frequency, odd_frequency])
plt.xlabel("Type")
plt.ylabel("Frequency")
plt.title("Even vs Odd Outcomes")
plt.show()
```

### 15. Create a chart showing experimental probability of Even, Odd, Greater than 4 and Less than 4.

```python
probabilities = [
    even_probability,
    odd_probability,
    probability_greater_4,
    probability_less_4
]

labels = ["Even", "Odd", "Greater than 4", "Less than 4"]

plt.bar(labels, probabilities)
plt.xlabel("Outcome")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probabilities")
plt.xticks(rotation=20)
plt.show()
```

## B) Probability Distributions

### 1. Generate 1,000 random numbers from a Normal distribution with Mean = 50 and Standard deviation = 10.

```python
normal_data = np.random.normal(50, 10, 1000)

print("First 10 values:")
print(normal_data[:10])
```

### 2. Calculate Mean, Median, Mode, Variance and Standard Deviation.

```python
print("Mean:", np.mean(normal_data))
print("Median:", np.median(normal_data))
print("Mode:", pd.Series(normal_data).mode()[0])
print("Variance:", np.var(normal_data))
print("Standard Deviation:", np.std(normal_data))
```

### 3. Calculate Mean, Median, Variance, Standard Deviation, Minimum and Maximum.

```python
print("Mean:", np.mean(normal_data))
print("Median:", np.median(normal_data))
print("Variance:", np.var(normal_data))
print("Standard Deviation:", np.std(normal_data))
print("Minimum:", np.min(normal_data))
print("Maximum:", np.max(normal_data))
```

### 4. Create a histogram of the Normal distribution and analyze its shape.

```python
plt.hist(normal_data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

```python
plt.boxplot(normal_data)
plt.ylabel("Values")
plt.title("Box Plot of Normal Distribution")
plt.show()
```

## C) Coin Toss Task

### 1. Simulate 1,000 coin tosses and calculate the experimental probability of Heads and Tails.

```python
tosses = np.random.choice(["Heads", "Tails"], 1000)

heads = np.sum(tosses == "Heads")
tails = np.sum(tosses == "Tails")

heads_probability = heads / 1000
tails_probability = tails / 1000

print("Number of Heads:", heads)
print("Number of Tails:", tails)
print("Experimental Probability of Heads:", heads_probability)
print("Experimental Probability of Tails:", tails_probability)
```

### 2. Calculate the theoretical probability of Heads and Tails.

```python
theoretical_heads = 1 / 2
theoretical_tails = 1 / 2

print("Theoretical Probability of Heads:", theoretical_heads)
print("Theoretical Probability of Tails:", theoretical_tails)
```

### 3. Create a bar chart showing the frequency of Heads and Tails.

```python
plt.bar(["Heads", "Tails"], [heads, tails])
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Heads and Tails")
plt.show()
```

### 4. Create a pie chart showing the percentage of Heads and Tails.

```python
plt.pie([heads, tails],
         labels=["Heads", "Tails"],
         autopct="%1.1f%%")

plt.title("Percentage of Heads and Tails")
plt.show()
```

### 5. If a coin is tossed 10 times and Heads occurs 8 times, calculate the experimental probability of Heads.

```python
heads = 8
total_tosses = 10

experimental_probability = heads / total_tosses

print("Experimental Probability of Heads:", experimental_probability)
```

### 6. Simulate 10,000 coin tosses and create a complete analysis.

```python
# Simulate 10000 coin tosses
tosses = np.random.choice(["Heads", "Tails"], 10000)

# Count Heads and Tails
heads = np.sum(tosses == "Heads")
tails = np.sum(tosses == "Tails")

# Experimental probabilities
experimental_heads = heads / 10000
experimental_tails = tails / 10000

# Theoretical probabilities
theoretical_heads = 0.5
theoretical_tails = 0.5

# Difference
difference_heads = abs(experimental_heads - theoretical_heads)
difference_tails = abs(experimental_tails - theoretical_tails)

# Display results
print("Total Tosses:", 10000)
print("Number of Heads:", heads)
print("Number of Tails:", tails)

print("\nExperimental Probability of Heads:", experimental_heads)
print("Experimental Probability of Tails:", experimental_tails)

print("\nTheoretical Probability of Heads:", theoretical_heads)
print("Theoretical Probability of Tails:", theoretical_tails)

print("\nDifference for Heads:", difference_heads)
print("Difference for Tails:", difference_tails)

# Bar Chart
plt.bar(["Heads", "Tails"], [heads, tails])
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Heads and Tails")
plt.show()

# Pie Chart
plt.pie([heads, tails],
        labels=["Heads", "Tails"],
        autopct="%1.1f%%")

plt.title("Percentage of Heads and Tails")
plt.show()
```
