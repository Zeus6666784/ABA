```python
from scipy.optimize import linprog

# 1. Decision variables
print("1. Decision Variables:")
print("x = Units of Product A")
print("y = Units of Product B")

# 2. Objective
print("\n2. Objective: Maximize Total Profit")

# 3. Objective function
c = [-40, -30]

# 4. Machine-hour constraint: 2x + y <= 100

# 5. Labour-hour constraint: x + 2y <= 80
A = [[2, 1], [1, 2]]
b = [100, 80]

# 6. Non-negative constraints
bounds = [(0, None), (0, None)]

# 7. Solve linear programming problem
result = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method="highs")

# 8. Optimal units
x = result.x[0]
y = result.x[1]
print("\n8. Optimal Units:")
print("Product A:", x)
print("Product B:", y)

# 9. Maximum profit
profit = 40 * x + 30 * y
print("\n9. Maximum Profit:", profit)

# 10. Check resource constraints
machine_used = 2 * x + y
labour_used = x + 2 * y
print("\n10. Resource Usage:")
print("Machine Hours Used:", machine_used)
print("Machine Hours Available:", 100)
print("Labour Hours Used:", labour_used)
print("Labour Hours Available:", 80)

# 11. Optimal solution
solution = {
    "Product A Units": x,
    "Product B Units": y,
    "Maximum Profit": profit,
    "Machine Hours Used": machine_used,
    "Labour Hours Used": labour_used
}
print("\n11. Optimal Solution:")
print(solution)

# 12. Business interpretation
print("\n12. Business Interpretation:")
print("The company should produce the optimal quantities of Product A and Product B to maximize profit.")

# 13. Conclusion
print("\n13. Conclusion:")
print(f"The maximum profit is {profit:.2f} with {x:.2f} units of Product A and {y:.2f} units of Product B.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# 14. Generate random demand for 30 days
demand = np.random.randint(10, 61, 30)

# 15. Display demand
print("15. Daily Demand:")
print(demand)

# 16. Inventory levels
inventory_levels = [30, 40, 50]

# 17-21. Simulate inventory
results = []

for inventory in inventory_levels:
    total_sold = 0
    total_shortage = 0

    for day, demand_value in enumerate(demand, 1):
        units_sold = min(demand_value, inventory)
        units_left = max(inventory - demand_value, 0)
        shortage = max(demand_value - inventory, 0)

        total_sold += units_sold
        total_shortage += shortage

        results.append([
            inventory, day, demand_value,
            units_sold, units_left, shortage
        ])

results_df = pd.DataFrame(
    results,
    columns=["Inventory", "Day", "Demand", "Units Sold", "Units Left", "Shortage"]
)

print("\n17-20. Simulation Results:")
print(results_df)

# 21. Total units sold and shortage
summary = results_df.groupby("Inventory").agg({
    "Units Sold": "sum",
    "Shortage": "sum"
}).reset_index()

print("\n21. Total Units Sold and Shortage:")
print(summary)

# 22 & 23. Store scenarios in DataFrame
print("\n22-23. Inventory Scenarios:")
print(summary)

# 24. Compare results
print("\n24. Comparison:")
print(summary)

# 25. Charts
plt.figure(figsize=(8, 5))
plt.bar(summary["Inventory"].astype(str), summary["Units Sold"])
plt.xlabel("Inventory Level")
plt.ylabel("Total Units Sold")
plt.title("Sales by Inventory Level")
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(summary["Inventory"].astype(str), summary["Shortage"])
plt.xlabel("Inventory Level")
plt.ylabel("Total Shortage")
plt.title("Shortage by Inventory Level")
plt.show()

# 26. Effect of inventory
print("\n26. Effect of Inventory:")
print("Increasing inventory generally increases units sold and reduces shortage.")

# 27. Conclusion
best_inventory = summary.loc[summary["Shortage"].idxmin(), "Inventory"]
print("\n27. Conclusion:")
print(f"Inventory level {best_inventory} results in the lowest shortage for the simulation.")
```
