# Practical 12: Analysis of Variance and Experimental Design

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway

# 1. Load dataset
df = pd.read_csv("Retail and wherehouse Sale - Retail and wherehouse Sale.csv")

# 2. Display first five records and dimensions
print("First Five Records:")
print(df.head())
print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# 3. Column names and data types
print("\nColumn Names:")
print(df.columns)
print("\nData Types:")
print(df.dtypes)

# 4. Missing values and duplicates
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Records:", df.duplicated().sum())

# 5. ITEM TYPE categories
print("\nITEM TYPE Categories:")
print(df["ITEM TYPE"].unique())

# 6. Average RETAIL SALES for each ITEM TYPE
average_sales = df.groupby("ITEM TYPE")["RETAIL SALES"].mean()
print("\n6. Average Retail Sales:")
print(average_sales)

# 7. Box plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="ITEM TYPE", y="RETAIL SALES")
plt.xticks(rotation=45)
plt.title("Retail Sales by Item Type")
plt.tight_layout()
plt.show()

# 8. Hypotheses
print("\n8. Hypotheses:")
print("H0: Mean retail sales are equal among all item types.")
print("H1: At least one item type has a different mean retail sales.")

# 9. One-way ANOVA
groups = [group["RETAIL SALES"].dropna() for _, group in df.groupby("ITEM TYPE")]
f_stat, p_value = f_oneway(*groups)

# 10. F-statistic and p-value
print("\n10. F-statistic:", f_stat)
print("P-value:", p_value)

# 11. Compare p-value
alpha = 0.05
print("\n11. Significance Level:", alpha)

# 12. Decision
if p_value < alpha:
    print("12. Decision: Reject H0")
else:
    print("12. Decision: Do not reject H0")

# 13. Conclusion
print("\n13. Conclusion:")
if p_value < alpha:
    print("There is a significant difference in retail sales among item types.")
else:
    print("There is no significant difference in retail sales among item types.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.formula.api import ols

# 14. Select factors and response
print("14. Factors: ITEM TYPE and YEAR")
print("Response: RETAIL SALES")

# 15. Average sales by ITEM TYPE and YEAR
average_combination = df.groupby(
    ["ITEM TYPE", "YEAR"]
)["RETAIL SALES"].mean().reset_index()

print("\n15. Average Retail Sales:")
print(average_combination)

# 16. Visualization
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=average_combination,
    x="YEAR",
    y="RETAIL SALES",
    hue="ITEM TYPE",
    marker="o"
)
plt.title("Retail Sales by Item Type Across Years")
plt.xlabel("Year")
plt.ylabel("Retail Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 17. Main effect of ITEM TYPE
item_type_effect = df.groupby("ITEM TYPE")["RETAIL SALES"].mean()
print("\n17. Main Effect of ITEM TYPE:")
print(item_type_effect)

# 18. Main effect of YEAR
year_effect = df.groupby("YEAR")["RETAIL SALES"].mean()
print("\n18. Main Effect of YEAR:")
print(year_effect)

# 19. Interaction effect
interaction_effect = df.groupby(
    ["ITEM TYPE", "YEAR"]
)["RETAIL SALES"].mean()

print("\n19. Interaction Effect:")
print(interaction_effect)

# 20. Factorial ANOVA
model = ols(
    'Q("RETAIL SALES") ~ C(Q("ITEM TYPE")) * C(YEAR)',
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

# 21. Statistical results and p-values
print("\n21. Factorial ANOVA Results:")
print(anova_table)

# 22. Identify significant effects
print("\n22. Significance Test:")
for effect, row in anova_table.iterrows():
    if row["PR(>F)"] < 0.05:
        print(effect, "has a significant effect.")
    else:
        print(effect, "does not have a significant effect.")

# 23. Final conclusion
print("\n23. Final Business Conclusion:")
significant_effects = anova_table[anova_table["PR(>F)"] < 0.05]

if len(significant_effects) > 0:
    print("The ANOVA indicates that at least one factor or interaction has a significant effect on retail sales.")
else:
    print("The ANOVA does not indicate a significant effect from the selected factors.")

print("\nLavkush T048")
```
