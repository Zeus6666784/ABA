## Q1

```
#Split Genres
genre=df['listed_in'].dropna().str.split(', ').explode()
```

```
#Top 10 Genres
top10=genre.value_counts().head(10)
print(top10)
```

```
#Bottom 10 Genres
bottom10=genre.value_counts().tail(10)
print(bottom10)
```

```
#Most Popular Genre
print(genre.value_counts().idxmax())
```

```
#Least Popular Genre
print(genre.value_counts().idxmin())
```

```
#Bar Chart
top10.plot(kind='bar',figsize=(8,5))
plt.title("Top 10 Genres")
plt.show()
```

## Q2

```
#Movies
movies=df[df['type']=='Movie']
tv=df[df['type']=='TV Show']
```

```
#Oldest Movie
print(movies.sort_values('release_year').head(1))
```

```
#Newest Movie
print(movies.sort_values('release_year',ascending=False).head(1))
```

```
#Oldest TV Show
print(tv.sort_values('release_year').head(1))
```

```
#Newest TV Show
print(tv.sort_values('release_year',ascending=False).head(1))
```

## Q3

```
#Split Countries
country=df['country'].dropna().str.split(', ').explode()
```

```
#Highest Content Country
print(country.value_counts().head(1))
```

```
#Top 10 Countries
top=country.value_counts().head(10)
print(top)
```

```
#Bottom 10 Countries
bottom=country.value_counts().tail(10)
print(bottom)
```

```
#Horizontal Bar Chart
top.sort_values().plot(kind='barh',figsize=(8,5))
plt.title("Top 10 Countries")
plt.show()
```

## Q4

```
#Movie Directors
movie_director=movies['director'].dropna().str.split(', ').explode()
```

```
#TV Show Directors
tv_director=tv['director'].dropna().str.split(', ').explode()
```

```
#Highest Movie Director
print(movie_director.value_counts().head(1))
```

```
#Top 10 Directors
topdir=df['director'].dropna().str.split(', ').explode().value_counts().head(10)
print(topdir)
```

```
#Bar Chart
topdir.plot(kind='bar',figsize=(8,5))
plt.title("Top 10 Directors")
plt.xticks(rotation=45)
plt.show()
```

## Q5

```
#Compare Releases
before=(df['release_year']<2020).sum()
during=(df['release_year']==2020).sum()
after=(df['release_year']>2020).sum()

release=pd.Series([before,during,after],
index=['Before 2020','During 2020','After 2020'])
print(release)
```

```
#Bar Chart
release.plot(kind='bar')
plt.title("Netflix Releases Comparison")
plt.show()
```
