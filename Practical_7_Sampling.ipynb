# Practical 7: Sampling and Sampling Distributions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 1. Select a real-life business topic
print("1. Business Topic: Customer Satisfaction with Online Shopping")

# 2. Prepare questionnaire
print("\n2. Questionnaire:")
print("Q1. Age")
print("Q2. Gender")
print("Q3. Monthly Online Shopping Spend")
print("Q4. Satisfaction Rating (1-10)")
print("Q5. Shopping Frequency per Month")

# 3 & 4. Generate and store 50 responses
data = pd.DataFrame({
    "Age": np.random.randint(18, 55, 50),
    "Gender": np.random.choice(["Male", "Female"], 50),
    "Monthly_Spend": np.random.randint(1000, 10000, 50),
    "Satisfaction": np.random.randint(5, 11, 50),
    "Shopping_Frequency": np.random.randint(1, 10, 50)
})
data.to_csv("survey_data.csv", index=False)

# 5. Display first five records
print("\n5. First Five Records:")
print(data.head())

# 6. Population and sample
print("\n6. Population: All online shoppers")
print("Sample: 50 selected respondents")

# 7. Sampling technique
print("\n7. Sampling Technique: Simple Random Sampling")

# 8. Numerical and categorical variables
print("\n8. Numerical Variables:")
print(["Age", "Monthly_Spend", "Satisfaction", "Shopping_Frequency"])
print("Categorical Variables:")
print(["Gender"])

# 9. Select numerical variable
variable = data["Monthly_Spend"]

# 10. Sample size, mean, median and standard deviation
print("\n10. Sample Size:", variable.size)
print("Mean:", variable.mean())
print("Median:", variable.median())
print("Standard Deviation:", variable.std())

# 11. Chart
plt.figure(figsize=(8, 5))
plt.hist(variable, bins=10)
plt.xlabel("Monthly Spend")
plt.ylabel("Frequency")
plt.title("Distribution of Monthly Online Shopping Spend")
plt.show()

# 12. Observations
print("\n12. Observations:")
print("1. Monthly spending varies among respondents.")
print("2. Most respondents have moderate monthly spending.")

# 13. Select numerical variable
selected_variable = data["Monthly_Spend"]

# 14. Random sample of 30 observations
sample = selected_variable.sample(30, random_state=42)

# 15. Calculate sample mean
print("\n15. Mean of Sample of 30:", sample.mean())

# 16 & 17. Repeat sampling 100 times
sample_means = []
for i in range(100):
    sample = selected_variable.sample(30)
    sample_means.append(sample.mean())
sample_means = np.array(sample_means)

# 18. First 10 sample means
print("\n18. First 10 Sample Means:")
print(sample_means[:10])

# 19. Mean of all sample means
print("\n19. Mean of Sample Means:", sample_means.mean())

# 20. Standard deviation of sample means
print("\n20. Standard Deviation of Sample Means:", sample_means.std())

# 21. Histogram
plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=10)
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of Sample Means")
plt.show()

# 22. Observe shape
print("\n22. Observation: The sampling distribution tends toward a bell-shaped distribution.")

# 23. Compare means
print("\n23. Original Mean:", selected_variable.mean())
print("Sampling Distribution Mean:", sample_means.mean())

# 24. Compare different sample sizes
sample_sizes = [10, 20, 30, 40]
distribution_std = []
for size in sample_sizes:
    means = []
    for i in range(100):
        means.append(selected_variable.sample(size).mean())
    distribution_std.append(np.std(means))

print("\n24. Sample Size vs Standard Deviation:")
for size, std in zip(sample_sizes, distribution_std):
    print(size, "->", std)

# 25. Standard error
standard_error = selected_variable.std() / np.sqrt(30)
print("\n25. Standard Error:", standard_error)

# 26. Central Limit Theorem
print("\n26. Central Limit Theorem:")
print("As the number of random samples increases, the distribution of sample means approaches a normal distribution.")

# 27. Estimate population mean
estimated_population_mean = sample_means.mean()
print("\n27. Estimated Population Mean:", estimated_population_mean)

# 28. Conclusion
print("\n28. Conclusion:")
print("The sampling distribution of sample means demonstrates the Central Limit Theorem.")
print("The mean of the sample means provides an estimate of the population mean.")

print("\nLavkush T048")
```
